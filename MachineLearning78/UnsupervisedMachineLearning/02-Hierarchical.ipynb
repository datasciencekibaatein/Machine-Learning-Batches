# Hierarchical Clustering

Hierarchical Clustering is one of the most popular unsupervised machine learning algorithm used to group similar observations together.

Unlike K-Means, Hierarchical Clustering does not require specifying specifying the number of clusters beforehand. Intead , It creates a hierarchy of clusters represented using a Dendrogram.

## Types of Hierarchical Clustering

1. Agglomerative Hierarchical Clustering (Bottom - to - up)
2. Divisive Clustering (Top - Down)

### Linkage Methods

After clusters are formed, we need a rule to measure distance between clusters.

This is called Linkage

#### 1. Single Linkage
Nearest neighbor.. Distance between clusters.(minimum distance)
#### 2. Complete linkage
Farthest neighbor.. Maximum distance(compact clusters)-- it is sensitive to outliers.
#### 3. Average Linkage
Average distance between clusters, Blananced approach, average distance = summation of distances / total pairs
#### 4. Ward Linkage
Most popular, minimizes increase in variance, produces compact clusters, usually best choice

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
```

```python
X = load_iris().data
```

Hierarchical clustering relies on distance so feature should be on same scale

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
from scipy.cluster.hierarchy import dendrogram,linkage

Z = linkage(
    X,
    method='ward'
)

plt.figure(figsize=(16,10))
dendrogram(Z)
plt.title("Dendrogram")
plt.xlabel("Sample")
plt.ylabel("Distance")
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

model = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

labels = model.fit_predict(X)
print(labels)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(
    X[:,0],
    X[:,1],
    c = labels
)
plt.xlabel("col1")
plt.ylabel("col2")
plt.title("Hierarchical clustering")
plt.show()
```
